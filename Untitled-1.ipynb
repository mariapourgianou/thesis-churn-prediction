```python
import pandas as pd
df = pd.read_csv('customer_model.csv')  
print(df.columns.tolist())
# εισαγωγη βιβλιοθηκης για την δημιουργια του μοντελου
# διαβαζει τα δεδομενα απο το αρχειο csv
# τυπωει τις 276 στηλες του αρχειου csv. βλεπω τι μεταβλητες εχει το αρχειο και μπορω να επιλεξω ποιες με ενδιαφερουν για την δημιουργια του μοντελου
```

```python
print(df.info()) # πληροφοριες για πληθος γραμμων, στηλες, τυπους δεδομενων και μη κενες τιμες
print(df.head()) # πρωτες 5 γραμμες του dataframe
```

```python
print(df['target_event'].value_counts()) # ποσες φορες εμφανιζεται το 1 και ποσες το 0 στην στηλη target_event. Αυτο ειναι σημαντικο για να δω αν τα δεδομενα ειναι ισορροπημενα. 
```

```python
# Έλεγχος μεγέθους και ελλιπών τιμών
print(df.shape)
print(df.isna().sum().sum()) # πλήθος ελλιπών τιμών

# Μία γραμμή ανά χρήστη;
print(df['user_id'].nunique())

# Ποιες στήλες είναι στόχοι
print([c for c in df.columns if c.startswith('target')]) # βρισκω οσες στηλες ξεκινανε με target. Αυτες ειναι οι στοχοι που θελω να προβλεψω. Στην περιπτωση μου ειναι η στηλη target_event

# Λεξικό δεδομένων
dd = pd.read_csv('data_dict.csv', encoding='cp1252')
print(dd.shape) # μεγεθος λεξικου δεδομενων
print(dd.head(30)) # οι πρωτες 30 γραμμες του λεξικου δεδομενων. Αυτο το αρχειο περιεχει πληροφοριες για τις στηλες του αρχειου customer_model.csv. Ειναι σημαντικο για να καταλαβω τι σημαινει καθε στηλη και να επιλεξω ποιες με ενδιαφερουν για την δημιουργια του μοντελου
```

```python
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', 250)
print(dd.columns.tolist()) #Τυπώνει τα ονόματα των 4 στηλών του λεξικού
print(dd.to_string()) # τυπωνει ολοκληρο το λεξικο 
```

```python
dd['Set'] = dd['Set'].ffill()
print(dd['Set'].value_counts()) # μετρα ποσες μεταβλητες εχει καθε ομαδα/κατηγορια. 
# Αυτο ειναι σημαντικο για να δω ποιες μεταβλητες ειναι σχετικες με το προβλημα που θελω να λυσω και ποιες οχι.
```

```python
print(df.groupby('user_id').size().describe()) #στατιστικά
print(df[df['user_id'] == df['user_id'].iloc[0]][['user_id', 'start_month_mean', 'target_event']])
# κρατω μονο τον πρωτο χρηστη και τυπωνω τις στηλες user_id, start_month_mean, target_event. 
# Αυτο το κανω για να δω αν οι τιμες της στηλης start_month_mean ειναι ιδιες για τον ιδιο χρηστη.
# Αν ειναι ιδιες μπορω να κρατησω μονο μια γραμμη ανα χρηστη και να μην εχω πολλες γραμμες με τον ιδιο χρηστη.
```

```python
rc = [c for c in df.columns if 'recency' in c]
print(df.groupby('target_event')[rc].mean())
# φτιαχνω λιστα με ολες τις στηλες που περιεχουν την λεξη recency και τυπωνω τον μεσο ορο τους για καθε τιμη της στηλης target_event.
#η κλάση 1 έχει σχεδόν διπλάσιο recency (περίπου 72 ημέρες έναντι 38), δηλαδή οι χρήστες της είχαν να εμφανιστούν πολύ περισσότερο καιρό.
# Αυτό είναι ισχυρή ένδειξη ότι η κλάση 1 είναι το churn.
```

```python
print('time_step' in df.columns)
print(df['time_step'].value_counts().sort_index())
print(df.groupby('user_id')['time_step'].nunique().describe())
#Πόσα διαφορετικά βήματα έχει ο κάθε χρήστης
# Αυτο το κομματι του κωδικα ειναι σημαντικο για να δω αν η στηλη time_step ειναι ιδια για τον ιδιο χρηστη.
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(figsize=(6, 4))
sns.countplot(x='target_event', data=df, ax=ax)
ax.set_xlabel('target_event (0 = όχι churn, 1 = churn)')
ax.set_ylabel('Πλήθος εγγραφών')
ax.set_title('Κατανομή της μεταβλητής-στόχου')
for p in ax.patches:
    ax.annotate(f"{int(p.get_height()):,}".replace(',', '.'),
                (p.get_x() + p.get_width() / 2, p.get_height()),
                ha='center', va='bottom')
plt.tight_layout()
plt.savefig('eikona_3_1.png', dpi=300)
plt.show()
```
